```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
stim_amps = np.load('stim_amplitudes.npy')
stim_durations = np.load('stim_durations.npy')
dur_mesh, amp_mesh = np.meshgrid(stim_durations, stim_amps)
areas = np.load('areas.npy')
```

```python
areas = areas.reshape(areas.shape[0], 50, 50)
```

```python
# import Patch
from matplotlib.patches import Patch
```

```python
plt.figure()
plt.pcolormesh(dur_mesh, amp_mesh, areas[0].reshape(50, 50), shading='auto', cmap='binary')
legend_elements = [Patch(facecolor='black', edgecolor='black', label='State-dependent')]
plt.legend(handles=legend_elements)
plt.ylabel(r'$I_{app}$')
plt.xlabel(r'$\tau_{dur} (s)$')
plt.yscale('log')
plt.show()
```
